Real Data

I selected two real biological repeated-count datasets that fit the AIM-DSPP framework particularly well.

**Dataset 1 — `MASS::epil`** contains 236 two-week seizure-count observations from 59 epilepsy patients. An eight-week pre-treatment seizure count is available before randomization, so the code uses:

$$R_i = \frac{\text{baseline seizures}_i}{4}$$

as the patient's externally observed two-week directing intensity. The original data analysis is Thall & Vail (1990), DOI 10.2307/2532086.

**Dataset 2 — `glmmTMB::Owls`** contains 599 repeated observations of sibling-negotiation/begging counts from barn-owl nestlings, together with nest, food treatment, provisioning-parent sex, arrival time, and brood size. The biological source is Roulin & Bersier (2007), DOI 10.1016/j.anbehav.2007.01.027.

Both are loaded first through `statsmodels.datasets.get_rdataset`, which is the official `statsmodels` mechanism for accessing datasets from R packages; a direct Rdatasets CSV route is included only as a fallback.
